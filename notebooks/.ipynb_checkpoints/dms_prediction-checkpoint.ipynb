```julia
using MultivariateStats
using DCAUtils
using LinearAlgebra
using PyPlot
using StatsBase
using SparseHop
using JLD2
```

## PF00014

```julia
Z = read_fasta_alignment("../../DataAttentionDCA/data/PF00014/PF00014_mgap6.fasta.gz", 0.9);
f1,f2 = compute_weighted_frequencies(Z, 22, 0.2);
```

```julia
L, M = size(Z);

correl_nat = f2 - f1*f1';
l, vv = eigen(correl_nat);

correl2 = zeros(21*L,21*L);
correl_h = []

for n in sortperm(abs.(l), rev = true)
    correl2 .+= l[n] .*vv[:,n].*vv[:,n]'
    push!(correl_h, cor(triu(correl_nat,21)[:], triu(correl2, 21)[:]))
end
 
plt.plot(correl_h)
plt.xlabel("# of eigenvectors used")
plt.ylabel("Spearman correlation")
plt.title("Natural Cij vs. in silico spectral decomposition")

println("Natural Cij vs Spectral decomposition (21 dims): $(correl_h[21])")
```
